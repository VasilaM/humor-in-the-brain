```python
from pathlib import Path

Path(reshaped_img_folder).mkdir(exist_ok=True)
```

```python
subs = [i for i in range(17, 83)]
exclude = [21, 22, 38, 56, 68, 69]
```

```python
def remove_subjects(sub_list, exclude_list):
    cleaned_list = sub_list
    for sub in exclude_list:
        if sub in sub_list:
            cleaned_list.remove(sub) 
    return cleaned_list
```

```python
subs = remove_subjects(subs, exclude)
```

```python
file_name = f"sub-300_task-piemanpni_space-MNI152NLin2009cAsym_res-native_desc-clean_bold.nii.gz"
file_dir = f"{cleaned_dir}sub-300/func/"
func_fn = file_dir + file_name
target_template_all = nib.load(func_fn)
print(target_template_all.shape)
target_template = nib.Nifti1Image(target_template_all.get_fdata()[:, :, :, 0], affine=target_template_all.affine)
```

```python
def reshape_data(subs):
    for sub in subs:
        file_name = f"sub-{sub:03d}_task-pieman_space-MNI152NLin2009cAsym_res-native_desc-clean_bold.nii.gz"
        file_dir = f"{cleaned_dir}sub-{sub:03d}/func/"
        func_fn = file_dir + file_name
        img = nib.load(func_fn)

        resampled_img = resample_to_img(img, target_template)
        print(resampled_img.shape)
        nib.save(resampled_img, reshaped_img_folder + f"sub-{sub:03d}_task-pieman_space-MNI152NLin2009cAsym_res-native_desc-clean_bold.nii.gz")
```

```python
reshape_data(subs)
```
